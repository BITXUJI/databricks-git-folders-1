```python
spark.read.format('csv') \
  .option('header', 'true') \
  .option('sep', ',') \
  .load('/Volumes/workspace/db_test/customer_test/customers-100.csv') \
  .createOrReplaceTempView('temp_bronze_customers')
```

```python
%sql
CREATE OR REPLACE TABLE db_test.bronze_customers AS
SELECT
  `Customer Id` AS customer_id,
  `First Name` AS first_name,
  `Last Name` AS last_name,
  Company AS company,
  City AS city,
  Country AS country,
  `Phone 1` AS phone_1,
  `Phone 2` AS phone_2,
  Email AS email,
  `Subscription Date` AS subscription_date,
  Website AS website
FROM temp_bronze_customers;
```

```python
%sql
CREATE OR REPLACE TABLE db_test.silver_customers AS
SELECT
   customer_id,
 first_name,
  last_name,
  company,
  city,
  country,
  phone_1,
  phone_2,
  email,
  CAST(subscription_date AS DATE) AS subscription_date,
  website
FROM db_test.bronze_customers;
```

```python
%sql
CREATE OR REPLACE TABLE db_test.gold_customers_subscription_made_by_year AS
SELECT
  YEAR(subscription_date) AS subscription_year,
  COUNT(*) AS customer_count
FROM db_test.silver_customers
GROUP BY YEAR(subscription_date)
ORDER BY subscription_year
```

```python
%sql
SELECT * FROM db_test.gold_customers_subscription_made_by_year;
```

```python
%sql
DROP TABLE IF EXISTS db_test.gold_customers_subscription_made_by_year;
```
