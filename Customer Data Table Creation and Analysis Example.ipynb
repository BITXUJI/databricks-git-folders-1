```python
%sql
CREATE OR REPLACE TABLE db_test.customers_sql
USING DELTA AS
SELECT
  Index AS index,
  `Customer Id` AS customer_id,
  `First Name` AS first_name,
  `Last Name` AS last_name,
  Company AS company,
  City AS city,
  Country AS country,
  `Phone 1` AS phone_1,
  `Phone 2` AS phone_2,
  Email AS email,
  `Subscription Date` AS subscription_date,
  Website AS website,
  _rescued_data AS rescued_data
FROM read_files(
  '/Volumes/workspace/db_test/customer_test/customers-100.csv',
  header => true
);
```

```python
table_full_name = "workspace.db_test.customers_sql"
table_df = spark.sql(f"SELECT * FROM {table_full_name}")
display(table_df)
```
